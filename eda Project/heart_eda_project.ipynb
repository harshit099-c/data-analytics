# ❤️ Heart Disease EDA Project

Exploratory Data Analysis on Heart Disease Dataset

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

## 📥 Load Dataset

```
df_heart = pd.read_csv("heart.csv")
df_heart.head()
```

## 🔍 Data Overview

```
df_heart.info()
df_heart.describe()
```

## 🧹 Data Cleaning

```
df_copy = df_heart.copy()

# Check null values
df_copy.isnull().sum()

# Remove duplicates
df_copy.drop_duplicates(inplace=True)

df_copy.shape
```

## 🔥 Correlation Heatmap

```
plt.figure(figsize=(8,10))
sns.heatmap(df_copy.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 📊 Categorical Feature Analysis

```
str_col = df_copy.select_dtypes(include='object').columns

for col in str_col:
    plt.figure(figsize=(6,4))
    ax = sns.countplot(x=df_copy[col])

    for container in ax.containers:
        ax.bar_label(container, label_type='edge', padding=3)

    plt.title(f"Count Plot of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

## 🔄 Encoding Categorical Variables

```
# Encode Sex
df_copy['Sex'] = df_copy['Sex'].str.strip().str.upper()
df_copy['Sex'] = df_copy['Sex'].replace({'M':1,'F':0})

# Rename for clarity
df_copy.rename(columns={'Sex':'isMale'}, inplace=True)

# One-hot encoding
df_copy = pd.get_dummies(df_copy, columns=['ChestPainType','RestingECG','ST_Slope'], dtype=int)

# Exercise Angina
df_copy['ExerciseAngina'] = df_copy['ExerciseAngina'].replace({'N':0,'Y':1})

df_copy.head()
```

## ⚙️ Feature Engineering

```
# Age categories
df_copy['Age_category'] = pd.cut(
    df_copy['Age'],
    bins=[28,40,52,64,float('inf')],
    labels=['Young','Mid','Senior','Old']
)

# Cholesterol range
df_copy['cholesterol_range'] = pd.cut(
    df_copy['Cholesterol'],
    bins=[0,200,240,600,float('inf')],
    labels=['Normal','Borderline','Risk','High']
)

# Ratio features
df_copy['BP_per_age'] = df_copy['RestingBP'] / df_copy['Age']
df_copy['HR_efficiency'] = df_copy['MaxHR'] / df_copy['Age']

# Risk Score
df_copy['RiskScore'] = (
    df_copy['Cholesterol'] +
    df_copy['RestingBP'] +
    df_copy['Oldpeak']
)

# Encode new categories
df_copy = pd.get_dummies(df_copy, columns=['Age_category','cholesterol_range'], dtype=int)

df_copy.head()
```

## 🔗 Feature vs Target Analysis

```
sns.countplot(x='isMale', hue='HeartDisease', data=df_copy)
plt.title("Gender vs Heart Disease")
plt.show()

sns.boxplot(x='HeartDisease', y='Age', data=df_copy)
plt.show()

sns.boxplot(x='HeartDisease', y='Cholesterol', data=df_copy)
plt.show()
```

## 🧠 Key Insights

- Older individuals have higher heart disease risk  
- High cholesterol and BP increase risk  
- Certain chest pain types strongly correlate with disease  
- Exercise-induced angina shows higher disease probability  
- Engineered features like RiskScore help in understanding patterns
