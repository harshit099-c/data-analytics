```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count=1
headers={
    'User-Agent':"Chrome/129.0.0.0 Safari/537.36"
}
while True:
    url=f"https://books.toscrape.com/catalogue/page-{page_count}.html"
    res=requests.get(url,headers=headers)
    res.encoding="utf-8"
    soup=BeautifulSoup(res.text,'lxml')

    title=soup.select(".product_pod h3 a")
    if not title:
        break

    with open(f"books_{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
    page_count=page_count+1
```

```python
about_books=[]
for i in range(1, page_count):
    with open(f"books_{i}.html", "r", encoding="utf-8") as f:
        html_content = f.read()
        soup2 = BeautifulSoup(html_content, "lxml")
        # Add code here to process soup2, e.g., extract data
        title_book=soup2.select_one("h3 a").get_text(strip=True)
        print(title_book)

        price=soup2.select_one(".product_price .price_color").get_text(strip=True)

        availability=soup2.select_one(".product_price .instock").get_text(strip=True)

        about_books.append([title_book,price,availability])
```

```python
import pandas as pd

df=pd.DataFrame(about_books,columns=["Title","price","availability"])
df.head()
```
