```python
from bs4 import BeautifulSoup

with open('web_scraped.html','r') as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml") #parse
soup

#methods

#1 find()

soup.find("h3")

#2 find_all()
country_name=soup.find_all("h3")
all_countries=[]
for i in country_name:
    print(i.get_text(strip=True)) #text inside element and strip to remove spaces

    #3 find_parent()
    print(i.find_parent("div")['class']) #to show attributes and parent class here class is attribute

    #4 find_next() and select_one()
    print(i.find_next("div").select_one("span.country-population").get_text(strip=True))

    #5 attrs
    print(i.attrs)
    print(i.get("class"))
    name=i.get_text(strip=True)
    population=i.find_next('div').select_one("span.country-population").get_text(strip=True)

    all_countries.append([name,population])
```

```python
import pandas as pd

df=pd.DataFrame(all_countries,columns=["Name","population"])
df

df.to_csv("cleaned.csv",index=Falsefrofromffffrrommmvvvv     )
```
