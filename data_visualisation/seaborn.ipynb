```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()

sns.set_theme(style="whitegrid")
tips=sns.load_dataset("tips")
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker",
    
)
```

```python
dow_jones=sns.load_dataset("dowjones")
print(dow_jones)
sns.lineplot(
    data=dow_jones,
    x="Date",
    y="Price",
    
)
sns.set_theme(style="ticks")
```

```python
#categorical plots

#1 bar plots

sns.barplot(
    data=tips,
    x="day",
    y="tip", #calculates avg tip on each day
    hue="sex"
)
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#distribution plots

penguin=sns.load_dataset("penguins")
sns.histplot(
    data=penguin,
    x="body_mass_g",
    bins=20,
    hue="sex",
    
)
```

```python
#matrix plot

#heatmaps
flights=sns.load_dataset("flights")
data_pivot=flights.pivot(index="month",columns="year",values="passengers")
data_pivot

sns.heatmap(
    data=data_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.tight_layout()
```

```python
fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    marker="o",
    errorbar=None
)

ax.set_title("avg billl value for each day")
fig.tight_layout()
```

```python
#1

penguin=sns.load_dataset("penguins")
penguin=penguin.dropna()
sns.scatterplot(
    data=penguin,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)

```

```python
#2
tips=sns.load_dataset("tips")
tips=tips.ffill(inplace=True)
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
medians=tips.groupby("day")["total_bill"].median()
print(max(medians))
plt.tight_layout()
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker"
)

plt.title("Q-4")
plt.tight_layout()
```

```python
#5
tip_pivot=tips.pivot_table(index="day",columns="time",values="tip")
sns.heatmap(
    data=tip_pivot,
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)

```
